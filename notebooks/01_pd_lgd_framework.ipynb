```python
import sys
sys.path.append("../src")

from feature_engineering import create_features
from pd_model import PDModel
from lgd_model import LGDModel
from expected_loss import calculate_expected_loss
from stress_testing import apply_stress
```

```python
import pandas as pd
df = pd.read_csv("../data/loans_sample.csv")

df.head()
```

```python
df = create_features(df)
df.head()
```

```python
pd_model = PDModel()

X_pd = df[["utilisation", "income_to_loan"]]
y_pd = df["default_flag"]

pd_model.fit(X_pd, y_pd)
pd_pred = pd.model.predict_proba(X_pd)

df["pd"] =pd_pred
df[["pd"]].head()
```

```python

lgd_model = LGDModel()

X_lgd = df[["utilisation"]]
y_lgd = df["loss_given_default"]

lgd_model.fit(X_lgd, y_lgd)
lgd_pred = lgd_model.predict(X_lgd)

df["lgd"] = lgd_pred
df[["lgd"]].head()
```

```python
df["expected_loss"] = calculate_expected_loss(
    df["pd"],
    df["lgd"],
    df["exposure"]
)

df[["pd", "lgd", "expected_loss"]].head()
```

```python
stressed_pd, stressed_lgd = apply_stress(
    df["pd"],
    df["lgd"],
    macro_factor=1.3
)

df["expected_loss_stressed"] = calculate_expected_loss(
    stressed_pd,
    stressed_lgd,
    df["exposure"]
)

df[["expected_loss", "expected_loss_stressed"]].head()
```
